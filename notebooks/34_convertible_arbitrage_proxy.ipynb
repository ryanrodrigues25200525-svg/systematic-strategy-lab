## 🧭 Context & methods

Convertible arbitrage seeks to exploit the relationship between a convertible bond’s bond floor, embedded equity option, credit risk, and the issuer’s stock. The actual hedge is security-specific and often uses short stock, options, financing, and dynamic delta management. CWB is only a long convertible-bond exposure, so this notebook labels the implementation gap clearly.

### 🔧 Key assumptions

- Asset: CWB as a liquid convertible-bond proxy; SPY is the benchmark.
- The notebook does not short individual stocks or dynamically hedge delta.
- Interest-rate, credit-spread, and equity-volatility effects remain embedded in the ETF price.
- The result is a proxy study, not a replication of a convertible-arbitrage hedge fund.

## ⚖️ What went well / what went wrong

**What went well:** The proxy helps a beginner see why convertible arbitrage is a hybrid credit/equity/option strategy rather than just “owning bonds.”

**What went wrong:** Without the short-equity hedge, the test still carries equity beta and can lose in credit stress. ETF duration, credit concentration, fees, and internal trading are not modeled.

**Research judgment:** Proxy labels matter: a positive result for CWB cannot be attributed to the arbitrage hedge that the ETF does not explicitly reproduce.

> The displayed metrics are sample-specific proxy results. They should be interpreted alongside costs, turnover, stress windows, and implementation gaps.
