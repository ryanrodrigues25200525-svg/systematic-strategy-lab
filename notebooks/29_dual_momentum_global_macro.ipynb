# 🧭 Dual momentum: global macro rotation

## 🧭 tl;dr

## 🧭 Context & methods

Dual momentum is a tactical global-allocation rule. Relative momentum chooses among risk assets; absolute momentum acts as a trend filter. The strategy is related to managed futures and global macro, but this implementation is long-only and ETF-based.

### 🔧 Key assumptions

- Risk assets: SPY, QQQ, IWM, EFA, and EEM.
- Defensive assets: TLT, GLD, and DBC.
- At each monthly rebalance, hold the strongest positive risk asset; otherwise hold the strongest positive defensive asset.
- No leverage, shorting, or cash yield is modeled.

## 📜 Strategy rules

1. Calculate each asset’s trailing 12-month return using only prior data.
2. If the strongest risk asset has positive momentum, hold it.
3. Otherwise, choose the strongest defensive asset with positive momentum.
4. Rebalance monthly and remain in cash when no candidate passes the absolute filter.

This rule is easy to understand but not immune to whipsaws or parameter sensitivity.

```python
weights = dual_momentum_weights(
    close,
    lookback=252,
    top_n=1,
    risk_assets=('SPY', 'QQQ', 'IWM', 'EFA', 'EEM'),
    defensive_assets=('TLT', 'GLD', 'DBC'),
)
```

## ⚖️ What went well / what went wrong

**What went well:** The two-stage rule clearly separates opportunity selection from risk control and gives the reader a direct way to study regime changes and defensive behavior.

**What went wrong:** A one-year lookback and a small ETF menu are arbitrary. The strategy can rotate after a large move has already happened, miss sharp rebounds, and behave differently when bonds and equities fall together.

**Research judgment:** Absolute momentum can reduce drawdowns in some samples, but it can also create long periods of cash or defensive exposure. That opportunity cost matters.

> The displayed metrics are sample-specific proxy results. They should be interpreted alongside costs, turnover, stress windows, and implementation gaps.
