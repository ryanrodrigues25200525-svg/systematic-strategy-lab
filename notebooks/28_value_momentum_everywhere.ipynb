## 🧭 Context & methods

Value compares an asset with a related alternative that looks expensive or cheap; momentum follows recent relative performance. Here, IWD versus IWF is used as a liquid equity value/growth spread, while momentum ranks a small cross-asset ETF universe. This is a practical proxy, not an exact eight-market replication.

### 🔧 Key assumptions

- Value sleeve: long IWD and short IWF at equal dollar risk.
- Momentum sleeve: monthly long the two strongest and short the two weakest assets across SPY, QQQ, EFA, EEM, TLT, GLD, and DBC.
- Value and momentum each contribute half of gross exposure.
- SPY is the benchmark; ETF fees, internal turnover, and financing are not fully modeled.

## 📜 Strategy rules

- **Value sleeve:** long the value ETF and short the growth ETF.
- **Momentum sleeve:** rank a small multi-asset universe by trailing return.
- **Portfolio:** combine both sleeves at equal gross risk and rebalance monthly.

Value and momentum are deliberately kept separate in the code so the reader can see whether diversification comes from the factor signals or merely from the ETF mix.

```python
def value_momentum_weights(panel, lookback=126, long_n=2, short_n=2):
    momentum_assets = ['SPY', 'QQQ', 'EFA', 'EEM', 'TLT', 'GLD', 'DBC']
    momentum_assets = [asset for asset in momentum_assets if asset in panel.columns]
    score = panel[momentum_assets].pct_change(lookback).shift(1)
    month_start = pd.Series(panel.index.to_period('M'), index=panel.index).ne(pd.Series(panel.index.to_period('M'), index=panel.index).shift(1)).to_numpy()
    candidates = pd.DataFrame(0.0, index=panel.index, columns=panel.columns)
    candidates['IWD'] = 0.25
    candidates['IWF'] = -0.25
    for timestamp in panel.index[month_start]:
        valid = score.loc[timestamp].dropna()
        if len(valid) < long_n + short_n:
            continue
        winners = valid.nlargest(long_n).index
        losers = valid.nsmallest(short_n).index
        candidates.loc[timestamp, winners] += 0.25 / len(winners)
        candidates.loc[timestamp, losers] -= 0.25 / len(losers)
    return candidates.ffill().fillna(0.0)

weights = value_momentum_weights(close)
```

## ⚖️ What went well / what went wrong

**What went well:** The blend turns an academic factor story into a portfolio-level experiment and tests whether two economically different signals diversify one another.

**What went wrong:** IWD/IWF is not a clean book-to-market factor, the asset universe is tiny, and the model ignores futures, currencies, stock-level shorts, leverage, and the paper’s global factor construction. Value and momentum can also suffer together during liquidity shocks.

**Research judgment:** A combined portfolio can look smoother without either sleeve being individually robust. Inspect the sleeves and the combined result separately.

> The displayed metrics are sample-specific proxy results. They should be interpreted alongside costs, turnover, stress windows, and implementation gaps.
